```python
import pandas as pd

# Give each column a name
columns = [
    "buying",
    "maint",
    "doors",
    "persons",
    "lug_boot",
    "safety",
    "class"
]

# Load the dataset
df = pd.read_csv("car.data", header=None, names=columns)
```

```python
# Display the first 10 cars
print(df.head(10))
```

```python
print(pd.crosstab(df["persons"], df["class"]))
```

```python
print(pd.crosstab(df["safety"], df["class"]))
```

```python
print(pd.crosstab(
    [df["buying"], df["safety"]],
    df["class"]
))
```

```python
# Count the classifications for each passenger capacity
pattern = (
    df.groupby("persons")["class"]
      .value_counts()
      .unstack(fill_value=0)
)

print(pattern)
```

```python
attributes = [
    "buying",
    "maint",
    "doors",
    "persons",
    "lug_boot",
    "safety"
]

for attribute in attributes:
    print(f"\nResults for {attribute}:")
    print(pd.crosstab(df[attribute], df["class"]))
```
