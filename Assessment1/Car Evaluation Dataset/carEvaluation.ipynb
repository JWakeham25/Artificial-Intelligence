```python
row = {"buying": "vhigh", "buying": "high", "buying": "med", "buying": "low",
       "maint": "vhigh", "maint": "high", "maint": "med", "maint": "low",
       "doors": 2, "doors": 3, "doors": 4, "doors": "5more",
       "persons": 2, "persons": 4, "persons": "more",
       "lug_boot": "small", "lug_boot": "med", "lug_boot": "big",
       "safety": "low", "safety": "med", "safety": "high"}

facts = {f"{attribute}={value}" for attribute, value in row.items()}
```

```python
rules = [
    ({"persons" : "2"}, "class = unacc"),
    ({"safety" : "low"}, "class = unacc"),
    ()
]
```

```python
def forward_chain_trace(facts, rules):
    known, trace = set(facts), []
    changed = True

    while changed:
        changed = False

        for conditions, conclusion in rules:
            if conditions <= known and conclusion not in known:
                known.add(conclusion)
                trace.append(
                    (sorted(conditions), conclusion)
                )
                changed = True

    return known, trace
```
