```python
row = {"buying": "vhigh", "maint": "vhigh", "doors": "2", "persons": "2", "lug_boot": "small", "safety": "low"}

facts = {f"{attribute} = {value}" for attribute, value in row.items()}
```

```python
rules = [
    ({"persons = 2"}, "class = unacc"),
    ({"safety = low"}, "class = unacc")
]
```

```python
def forward_chain_trace(facts, rules):
    known, trace = set(facts), []
    changed = True
    
    while changed:
        changed = False
        
        for conditions, conclusion in rules:
            if conditions <= known and conclusion not in known:
                known.add(conclusion)
                trace.append((sorted(conditions), conclusion))
                changed = True
                
    return known, trace

known_facts, trace = forward_chain_trace(facts, rules)
for conditions, conclusion in trace:
    print(f"{' + '.join(conditions)} => {conclusion}")
```
