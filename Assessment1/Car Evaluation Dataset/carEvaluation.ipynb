```python
def forward_chain_trace(facts, rules):
    known, trace = set(facts), []
    changed = True
    
    while changed:
        changed = False
        
        for conditions, conclusion in rules:
            if conditions <= known and conclusion not in known:
                known.add(conclusion)
                trace.append((sorted(conditions), conclusion))
                changed = True
                
    return known, trace

known_facts, trace = forward_chain_trace(facts, rules)
for conditions, conclusion in trace:
    print(f"{' + '.join(conditions)} => {conclusion}")
```

```python
def predict_car(row, rules):
    facts = {
        f"{attribute} = {value}"
        for attribute, value in row.items()
        if attribute != "class"
    }

    known_facts, trace = forward_chain_trace(facts, rules)

    predictions = {
        fact.split(" = ", 1)[1]
        for fact in known_facts
        if fact.startswith("class = ")
    }

    if len(predictions) == 1:
        prediction = predictions.pop()
    elif len(predictions) == 0:
        prediction = "unclassified"
    else:
        prediction = "conflict"

    return prediction, trace
```

```python
import pandas as pd

columns = ["buying", "maint", "doors", "persons", "lug_boot", "safety", "class"]

df = pd.read_csv("car.data", header=None, names=columns)

rows = df.to_dict("records")

print(f"Number of cars: {len(rows)}")
print(rows[0])  # Display the first row for verification
```

```python
def classify(row):
    prediction, trace = predict_car(row, rules)
    return prediction
```

```python
from collections import Counter

# rows is a list of dicts, each with a "class" key
majority = Counter(r["class"] for r in rows).most_common(1)[0][0]
baseline = sum(r["class"] == majority for r in rows) / len(rows)
print(f"Always predict {majority!r}: {baseline:.2f}")

accuracy = sum(classify(r) == r["class"] for r in rows) / len(rows)
print(f"My rules: {accuracy:.2f}")
```

```python
confusion = Counter((r["class"], classify(r)) for r in rows)

def precision_recall(rows, classify, cls):
    tp = sum(1 for r in rows if classify(r) == cls and r["class"] == cls)
    fp = sum(1 for r in rows if classify(r) == cls and r["class"] != cls)
    fn = sum(1 for r in rows if classify(r) != cls and r["class"] == cls)
    precision = tp / (tp + fp) if tp + fp else 0.0
    recall = tp / (tp + fn) if tp + fn else 0.0
    return precision, recall
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

y_true = [r["class"] for r in rows]
y_pred = [classify(r) for r in rows]

print(f"Accuracy: {accuracy_score(y_true, y_pred):.2f}")
print(classification_report(y_true, y_pred, zero_division=0))
print(confusion_matrix(y_true, y_pred, labels=sorted(set(y_true))))
```
